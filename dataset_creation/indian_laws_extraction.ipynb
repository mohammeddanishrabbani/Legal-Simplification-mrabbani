```python

import csv

with open('indian_laws_v2_breakdown.csv', 'w') as f:
    writer = csv.writer(f)
    writer.writerow(['legal_text', 'simplified_text'])
   
    for act in df['act_title'].unique():
        act_string = ""
        act_df = df[df['act_title'] == act]
        #convert the section column to int and sort the dataframe based on section
        act_df['section'] = act_df['section'].astype(int)
        act_df = act_df.sort_values('section')
        act_string = f"{act}\n\n"
        for index, act in act_df.iterrows():
            legal_text =  f"SECTION {act['section']} of {act['act_title']}\n{act['law']}"
            writer.writerow([legal_text, ""])

print("Done writing to csv")

```
