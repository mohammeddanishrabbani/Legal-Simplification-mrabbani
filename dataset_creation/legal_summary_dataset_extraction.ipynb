```python
from datasets import load_dataset

ds = load_dataset("lighteval/legal_summarization", "MultiLexSum")
```

```python
print(ds)
```

```python
import csv

with open("legal_summary_dataset.csv", "w") as f:
    writer = csv.writer(f)
    writer.writerow(["legal_text", "simplified_text"])
    for item in ds["train"]:
        writer.writerow([item["article"], ""])
    for item in ds["validation"]:
        writer.writerow([item["article"], ""])
    for item in ds["test"]:
        writer.writerow([item["article"], ""])



```
