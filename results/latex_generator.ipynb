```python
df = pd.read_csv("results/standard_metrics.csv")
```

```python
#delete last 3 columns
df = df.iloc[:, :-2]
df.head()
```

```python
def bold_max(s):
    is_max = s == s.max()
    return ['\\textbf{' + str(v) + '}' if m and type(m)!=str else str(v) for v, m in zip(s, is_max)]
```

```python
latex_df = df.apply(bold_max)
```

```python
latex_df.to_latex("results/OTHER.tex", index=False, escape=False, float_format="%.2f", column_format="lccccccccc", multirow=True, multicolumn=True,)
```
