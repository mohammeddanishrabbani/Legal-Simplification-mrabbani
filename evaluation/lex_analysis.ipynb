```python
import pandas as pd
import os
os.chdir("/home/danish/Legal-Simplification-mrabbani")
```

```python
dataset_path = "dataset/merged_dataset.csv"
```

```python
df = pd.read_csv(dataset_path)
```

```python
results = {
    "word_count_legal_text": [],
    "word_count_simplified_text": [],
    "word_count_difference": [],
    
}
```

```python
for row_num,row in df.iterrows():

    # Extract row data
    # print(row)
    legal_text = row["legal_text"]
    simplified_text = row["simplified_text"]
    
    # Count words in legal text
    word_count_legal_text = len(legal_text.split())
    
    # Count words in simplified text
    word_count_simplified_text = len(simplified_text.split())
    
    # Calculate difference
    word_count_difference = abs(word_count_legal_text - word_count_simplified_text)
    
    # Append results
    results["word_count_legal_text"].append(word_count_legal_text)
    results["word_count_simplified_text"].append(word_count_simplified_text)
    results["word_count_difference"].append(word_count_difference)
```

```python
results_df = pd.DataFrame(results)
```
